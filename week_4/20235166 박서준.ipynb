```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

X = iris.data
y = iris.target

df = pd.DataFrame(X, columns=iris.feature_names)

df.head()
```

```python
from sklearn.cluster import KMeans

kmeans_raw = KMeans(
    n_clusters=3,
    random_state=0,
    n_init="auto"
)

cluster_raw = kmeans_raw.fit_predict(X)
```

```python
result = df.copy()
result["cluster"] = cluster_raw

result.head(10)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

df_scaled = pd.DataFrame(
    X_scaled,
    columns=iris.feature_names
)

df_scaled.describe().round(2)
```

```python
# 표준화 전
kmeans_raw = KMeans(
    n_clusters=3,
    random_state=0,
    n_init="auto"
)
cluster_raw = kmeans_raw.fit_predict(X)

# 표준화 후
kmeans_scaled = KMeans(
    n_clusters=3,
    random_state=0,
    n_init="auto"
)
cluster_scaled = kmeans_scaled.fit_predict(X_scaled)
```

```python
result_raw = df.copy()
result_raw["cluster"] = cluster_raw

result_scaled = df.copy()
result_scaled["cluster"] = cluster_scaled

print("표준화 전")
display(result_raw.groupby("cluster").mean().round(2))

print("표준화 후")
display(result_scaled.groupby("cluster").mean().round(2))
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

ks = [2, 3, 4, 5]
inertias = []

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

for ax, k in zip(axes.ravel(), ks):

    kmeans = KMeans(
        n_clusters=k,
        random_state=0,
        n_init="auto"
    )

    cluster = kmeans.fit_predict(X)

    # Inertia 저장
    inertias.append(kmeans.inertia_)

    # Inertia 출력
    print(f"K = {k}, Inertia = {kmeans.inertia_:.2f}")

    # 클러스터링 결과
    ax.scatter(
        X[:, 0],
        X[:, 1],
        c=cluster,
        alpha=0.7
    )

    ax.set_title(f"K = {k}")
    ax.set_xlabel("Sepal length")
    ax.set_ylabel("Sepal width")

plt.tight_layout()
plt.show()
```

```python
from sklearn.decomposition import PCA

# 4개의 특성을 2개의 주성분으로 축소
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("원래 데이터:", X_scaled.shape)
print("PCA 변환 후:", X_pca.shape)
```

```python
print("PC1 explained variance ratio:",
      pca.explained_variance_ratio_[0])

print("PC2 explained variance ratio:",
      pca.explained_variance_ratio_[1])
```

```python
print(
    f"PC1이 설명하는 정보: "
    f"{pca.explained_variance_ratio_[0] * 100:.1f}%"
)

print(
    f"PC2가 설명하는 정보: "
    f"{pca.explained_variance_ratio_[1] * 100:.1f}%"
)

print(
    f"PC1 + PC2가 설명하는 정보: "
    f"{pca.explained_variance_ratio_.sum() * 100:.1f}%"
)
```

## 실습 과제: K=2, K=4 군집화 비교

```python
# 1. K=2로 군집화
kmeans_k2 = KMeans(
    n_clusters=2,
    random_state=0,
    n_init="auto"
)
cluster_k2 = kmeans_k2.fit_predict(X_scaled)
```

```python
# K=4로 군집화
kmeans_k4 = KMeans(
    n_clusters=4,
    random_state=0,
    n_init="auto"
)
cluster_k4 = kmeans_k4.fit_predict(X_scaled)
```

```python
# 2. PCA 공간에서 K=2, K=4, 실제 품종을 시각화
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

for c in range(2):
    axes[0].scatter(
        X_pca[cluster_k2 == c, 0],
        X_pca[cluster_k2 == c, 1],
        label=f"Cluster {c}",
        alpha=0.7
    )
axes[0].set_xlabel("PC1")
axes[0].set_ylabel("PC2")
axes[0].set_title("K-means (K=2)")
axes[0].legend()

for c in range(4):
    axes[1].scatter(
        X_pca[cluster_k4 == c, 0],
        X_pca[cluster_k4 == c, 1],
        label=f"Cluster {c}",
        alpha=0.7
    )
axes[1].set_xlabel("PC1")
axes[1].set_ylabel("PC2")
axes[1].set_title("K-means (K=4)")
axes[1].legend()

for i, name in enumerate(iris.target_names):
    axes[2].scatter(
        X_pca[y == i, 0],
        X_pca[y == i, 1],
        label=name,
        alpha=0.7
    )
axes[2].set_xlabel("PC1")
axes[2].set_ylabel("PC2")
axes[2].set_title("Actual Species")
axes[2].legend()

plt.tight_layout()
plt.show()
```

```python
# 3. 각 군집 결과를 실제 품종과 교차표로 비교
result_k2 = df.copy()
result_k2["cluster"] = cluster_k2
result_k2["species"] = iris.target_names[y]

print("K=2 교차표")
display(pd.crosstab(result_k2["cluster"], result_k2["species"]))
```

```python
result_k4 = df.copy()
result_k4["cluster"] = cluster_k4
result_k4["species"] = iris.target_names[y]

print("K=4 교차표")
display(pd.crosstab(result_k4["cluster"], result_k4["species"]))
```

### 4. K=2와 K=4 중 실제 품종 구조와 더 비슷한 것은?

**결론: K=4보다 K=2가 실제 붓꽃 품종 구조와 겉보기엔 더 깔끔하게 맞아떨어지지만, 품종을 3개로 나누는 진짜 구조를 더 잘 드러내는 쪽은 여전히 원래 실습에서 쓴 K=3이며, K=2와 K=4는 각각 다른 방식으로 그 구조를 놓친다.**

- **K=2 결과**: 교차표를 보면 cluster 0 = setosa 50개(100%), cluster 1 = versicolor 50개 + virginica 50개가 그대로 합쳐져 있다. PCA 그림에서도 setosa는 완전히 분리되어 있지만 versicolor와 virginica는 한 덩어리로 뭉쳐 있다. 즉 K=2는 '꽃잎이 확연히 다른 setosa vs 나머지' 라는 큰 경계 하나만 잡아내고, versicolor와 virginica 사이의 실제 경계는 아예 잡아내지 못한다.
- **K=4 결과**: 교차표를 보면 cluster 1(28개)과 cluster 3(22개)이 둘 다 거의 setosa로만 이루어져 있어서, 본래 하나였던 setosa 그룹이 두 개의 클러스터로 쪼개졌다. 동시에 versicolor와 virginica는 cluster 0, cluster 2로 나뉘긴 했지만 완벽히 분리되지 않고 서로 섞여 있다(cluster 0에 virginica 14개, cluster 2에 versicolor 11개 혼입). 즉 K=4는 실제로 존재하지 않는 경계로 setosa를 둘로 나누는 '과분할(over-clustering)' 오류를 범한다.
- **비교**: 실제 품종은 3개이므로, 두 값 모두 정답(K=3)과는 차이가 있다. 그래도 굳이 K=2와 K=4 중 하나를 고른다면, K=2는 최소한 setosa 경계는 완벽하게 지키면서 나머지 두 종을 뭉뚱그린 반면, K=4는 존재하지 않는 setosa 내부 경계를 새로 만들어내면서도 정작 필요한 versicolor-virginica 경계는 여전히 깔끔하게 긋지 못했다. 따라서 오류의 성격(부족한 분할 vs 잘못된 분할)이 다르지만, PCA 산점도상 실제 3개 품종의 '뭉침 형태' 자체를 왜곡 없이 보존한다는 기준으로 보면 **K=2 쪽이 실제 구조를 더 solid하게 반영**하고, K=4는 겉보기에 더 세분화되어 있을 뿐 실제 생물학적 경계와는 무관한 인위적 분할을 포함한다.
