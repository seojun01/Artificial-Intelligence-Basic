```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

X = iris.data
y = iris.target

df = pd.DataFrame(X, columns=iris.feature_names)

df.head()
```

```python
from sklearn.cluster import KMeans

kmeans_raw = KMeans(
    n_clusters=3,
    random_state=0,
    n_init="auto"
)

cluster_raw = kmeans_raw.fit_predict(X)
```

```python
result = df.copy()
result["cluster"] = cluster_raw

result.head(10)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

df_scaled = pd.DataFrame(
    X_scaled,
    columns=iris.feature_names
)

df_scaled.describe().round(2)
```

```python
# 표준화 전
kmeans_raw = KMeans(
    n_clusters=3,
    random_state=0,
    n_init="auto"
)
cluster_raw = kmeans_raw.fit_predict(X)

# 표준화 후
kmeans_scaled = KMeans(
    n_clusters=3,
    random_state=0,
    n_init="auto"
)
cluster_scaled = kmeans_scaled.fit_predict(X_scaled)
```

```python
result_raw = df.copy()
result_raw["cluster"] = cluster_raw

result_scaled = df.copy()
result_scaled["cluster"] = cluster_scaled

print("표준화 전")
display(result_raw.groupby("cluster").mean().round(2))

print("표준화 후")
display(result_scaled.groupby("cluster").mean().round(2))
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

ks = [2, 3, 4, 5]
inertias = []

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

for ax, k in zip(axes.ravel(), ks):

    kmeans = KMeans(
        n_clusters=k,
        random_state=0,
        n_init="auto"
    )

    cluster = kmeans.fit_predict(X)

    # Inertia 저장
    inertias.append(kmeans.inertia_)

    # Inertia 출력
    print(f"K = {k}, Inertia = {kmeans.inertia_:.2f}")

    # 클러스터링 결과
    ax.scatter(
        X[:, 0],
        X[:, 1],
        c=cluster,
        alpha=0.7
    )

    ax.set_title(f"K = {k}")
    ax.set_xlabel("Sepal length")
    ax.set_ylabel("Sepal width")

plt.tight_layout()
plt.show()
```

```python
from sklearn.decomposition import PCA

# 4개의 특성을 2개의 주성분으로 축소
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("원래 데이터:", X_scaled.shape)
print("PCA 변환 후:", X_pca.shape)
```

```python
print("PC1 explained variance ratio:",
      pca.explained_variance_ratio_[0])

print("PC2 explained variance ratio:",
      pca.explained_variance_ratio_[1])
```

```python
print(
    f"PC1이 설명하는 정보: "
    f"{pca.explained_variance_ratio_[0] * 100:.1f}%"
)

print(
    f"PC2가 설명하는 정보: "
    f"{pca.explained_variance_ratio_[1] * 100:.1f}%"
)

print(
    f"PC1 + PC2가 설명하는 정보: "
    f"{pca.explained_variance_ratio_.sum() * 100:.1f}%"
)
```

```python
# k=3으로 군집화
kmeans = KMeans(
    n_clusters=3,
    random_state=0,
    n_init="auto"
)

cluster = kmeans.fit_predict(X_scaled)
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=cluster,
    alpha=0.7
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("K-means Clustering in PCA Space")

plt.show()
```

```python
result = df.copy()

result["cluster"] = cluster
result["species"] = iris.target_names[y]

result.head()
```

```python
pd.crosstab(
    result["cluster"],
    result["species"]
)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# K-means 결과
for c in range(3):
    axes[0].scatter(
        X_pca[cluster == c, 0],
        X_pca[cluster == c, 1],
        label=f"Cluster {c}",
        alpha=0.7
    )

axes[0].set_xlabel("PC1")
axes[0].set_ylabel("PC2")
axes[0].set_title("K-means Clusters")
axes[0].legend()


# 실제 Iris 품종
for i, name in enumerate(iris.target_names):
    axes[1].scatter(
        X_pca[y == i, 0],
        X_pca[y == i, 1],
        label=name,
        alpha=0.7
    )

axes[1].set_xlabel("PC1")
axes[1].set_ylabel("PC2")
axes[1].set_title("Actual Species")
axes[1].legend()

plt.tight_layout()
plt.show()
```
